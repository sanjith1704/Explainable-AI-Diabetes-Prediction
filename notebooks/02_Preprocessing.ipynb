```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/diabetes.csv")

df.head()
```

```python
df.head()
```

```python
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
invalid_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

X[invalid_zero_columns] = X[invalid_zero_columns].replace(0, np.nan)

print("Missing values after replacing invalid zeros:")
print(X.isnull().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

print("Imputation completed successfully!")
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

print("Feature scaling completed successfully!")
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X.columns,
    index=X_test.index
)

X_train_scaled.head()
```

```python
print("Missing values in training data:")
print(X_train_scaled.isnull().sum())

print("\nMissing values in testing data:")
print(X_test_scaled.isnull().sum())
```

```python
print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

print("\nTraining target percentage:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting target percentage:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
print("Training feature means:")
print(X_train_scaled.mean().round(2))

print("\nTraining feature standard deviations:")
print(X_train_scaled.std().round(2))
```

```python
import joblib

joblib.dump(imputer, "../models/imputer.pkl")
joblib.dump(scaler, "../models/scaler.pkl")

print("Preprocessing objects saved successfully!")
```

```python
X_train_scaled.to_csv("../data/X_train.csv", index=False)
X_test_scaled.to_csv("../data/X_test.csv", index=False)

y_train.to_csv("../data/y_train.csv", index=False)
y_test.to_csv("../data/y_test.csv", index=False)

print("Processed datasets saved successfully!")
```
