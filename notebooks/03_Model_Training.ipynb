```python
import pandas as pd
import numpy as np
import joblib

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

print("Libraries imported successfully!")
```

```python
X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
logistic_model = LogisticRegression(
    random_state=42,
    max_iter=1000
)

random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

xgboost_model = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
)

print("Models created successfully!")
```

```python
logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
random_forest_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
xgboost_model.fit(X_train, y_train)

print("XGBoost trained successfully!")
```

```python
y_pred_lr = logistic_model.predict(X_test)
y_pred_rf = random_forest_model.predict(X_test)
y_pred_xgb = xgboost_model.predict(X_test)

print("Predictions generated successfully!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

models = {
    "Logistic Regression": (logistic_model, y_pred_lr),
    "Random Forest": (random_forest_model, y_pred_rf),
    "XGBoost": (xgboost_model, y_pred_xgb)
}

results = []

for name, (model, predictions) in models.items():

    probabilities = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1 Score": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

metrics_df = pd.DataFrame(results)

metrics_df
```

```python
metrics_df.to_csv("../results/model_metrics.csv", index=False)

print("Model metrics saved successfully!")
```

```python
joblib.dump(logistic_model, "../models/logistic_regression.pkl")
joblib.dump(random_forest_model, "../models/random_forest.pkl")
joblib.dump(xgboost_model, "../models/xgboost.pkl")

print("All models saved successfully!")
```

```python
import pandas as pd
import numpy as np
import joblib

X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()

print("Training and test data loaded successfully!")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from xgboost import XGBClassifier

xgboost_model = XGBClassifier(
    n_estimators=100,
    max_depth=3,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
)

print("XGBoost model created successfully!")
```

```python
xgboost_model.fit(X_train, y_train)

print("XGBoost trained successfully!")
```

```python
y_pred_xgb = xgboost_model.predict(X_test)
y_prob_xgb = xgboost_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
print("First 10 predictions:", y_pred_xgb[:10])
print("First 10 probabilities:", y_prob_xgb[:10])
```

```python
joblib.dump(xgboost_model, "../models/xgboost.pkl")

print("XGBoost model saved successfully!")
```

```python
scaler = joblib.load("../models/scaler.pkl")

print("Scaler loaded successfully!")
```

```python
patient = X_test.iloc[0].copy()

print("Testing glucose effect with scaling:\n")

for glucose in [70, 100, 130, 160, 190, 210]:

    test_patient = patient.copy()
    test_patient["Glucose"] = glucose

    # Convert to DataFrame
    patient_df = pd.DataFrame([test_patient])

    # Apply the same scaler used during training
    patient_scaled = scaler.transform(patient_df)

    # Convert back to DataFrame with feature names
    patient_scaled_df = pd.DataFrame(
        patient_scaled,
        columns=X_train.columns
    )

    probability = xgboost_model.predict_proba(
        patient_scaled_df
    )[0][1]

    print(f"Glucose = {glucose:3} → {probability:.2%}")
```
